##Data Transformation

```python
discrete_features = ['PericardialEffusion']
continuous_features = data.drop(columns=['PericardialEffusion', 'StillAlive']).columns
label = ['StillAlive']

print(continuous_features)
```

```python
data.shape
```

```python
for feature in discrete_features:
    data[feature] = data[feature].fillna(data[feature].mode()[0])
```

```python
for feature in continuous_features:
    data.boxplot(feature)
    plt.title(feature)
    plt.show()
```

## Data Transformation and Data Discretization

```python
features_with_outliers = ['WallMotionScore', 'WallMotionIndex', 'Mult']
```

```python
for feature in continuous_features:
    if feature in features_with_outliers:
        data[feature] = data[feature].fillna(data[feature].median())
    else:
        data[feature] = data[feature].fillna(data[feature].mean())
```

```python
data.shape
```

```python
from sklearn.neighbors import LocalOutlierFactor
lof = LocalOutlierFactor()
outliers_rows = lof.fit_predict(data)
```

```python
mask = outliers_rows != -1
```

```python
data.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
data = data[mask]
```

```python
# One-hot encoding untuk variabel kategori (PericardialEffusion)
data_encoded = pd.get_dummies(data, columns=discrete_features, drop_first=True)
```

```python
data.head()
```

```python
data
```

```python
# Memisahkan target dan fitur dari data yang sudah di-encode
X = data_encoded.drop(columns=['StillAlive'])
y = data_encoded['StillAlive']
```

```python
data.shape
```

##Data Processing

```python
from sklearn.preprocessing import StandardScaler

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=0)

# Standarisasi fitur numerik setelah split untuk mencegah Data Leakage
scaler = StandardScaler()

# Kita transformasikan kolom-kolom kontinu pada set Train dan Test
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[continuous_features] = scaler.fit_transform(X_train[continuous_features])
X_test_scaled[continuous_features] = scaler.transform(X_test[continuous_features])

# Untuk kemudahan model-model berikutnya, kita timpa X_train & X_test dengan versi yang sudah di-scale
X_train = X_train_scaled
X_test = X_test_scaled

print("Data Splitting dan Standarisasi berhasil diselesaikan dengan benar!")
```

```python
accuracy = {}
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.linear_model import LogisticRegression

# 1. Pastikan data bersih didefinisikan (tanpa kebocoran target 'Survival')
X_clean = data_encoded.drop(columns=['StillAlive', 'Survival'], errors='ignore')
y_clean = data_encoded['StillAlive']

# Tentukan ulang kolom numerik
continuous_features_clean = [col for col in continuous_features if col != 'Survival']
outlier_cols = ['WallMotionScore', 'WallMotionIndex', 'Mult']
normal_cols = [col for col in continuous_features_clean if col not in outlier_cols]

# 2. Definisikan kembali preprocessor
preprocessor = ColumnTransformer(
    transformers=[
        ('num_median', Pipeline([
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())
        ]), outlier_cols),
        ('num_mean', Pipeline([
            ('imputer', SimpleImputer(strategy='mean')),
            ('scaler', StandardScaler())
        ]), normal_cols)
    ],
    remainder='passthrough'
)

# 3. Definisikan cross-validation generator
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=0)

# Evaluasi Model 1: Logistic Regression menggunakan Pipeline dan CV
model1 = LogisticRegression(max_iter=200)

pipeline_lr = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', model1)
])

scores_lr = cross_val_score(pipeline_lr, X_clean, y_clean, cv=cv, scoring='accuracy')
print(f"Logistic Regression -> Rata-rata Akurasi: {scores_lr.mean()*100:.2f}% (Std: {scores_lr.std()*100:.2f}%)")
accuracy['Logistic Regression (CV)'] = scores_lr.mean() * 100
```

```python
# Evaluasi Model 2: Decision Tree menggunakan Pipeline dan CV
model2 = DecisionTreeClassifier(max_depth=3, random_state=0)

pipeline_dt = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', model2)
])

scores_dt = cross_val_score(pipeline_dt, X_clean, y_clean, cv=cv, scoring='accuracy')
print(f"Decision Tree -> Rata-rata Akurasi: {scores_dt.mean()*100:.2f}% (Std: {scores_dt.std()*100:.2f}%)")
accuracy['Decision Tree (CV)'] = scores_dt.mean() * 100
```

```python
# Evaluasi Model 3: Random Forest menggunakan Pipeline dan CV
model3 = RandomForestClassifier(max_depth=6, random_state=0)

pipeline_rf = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', model3)
])

scores_rf = cross_val_score(pipeline_rf, X_clean, y_clean, cv=cv, scoring='accuracy')
print(f"Random Forest -> Rata-rata Akurasi: {scores_rf.mean()*100:.2f}% (Std: {scores_rf.std()*100:.2f}%)")
accuracy['Random Forest (CV)'] = scores_rf.mean() * 100
```

```python
# Evaluasi Model 4: Gradient Boosting menggunakan Pipeline dan CV
model4 = GradientBoostingClassifier(n_estimators=300, learning_rate=0.1, random_state=0)

pipeline_gb = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', model4)
])

scores_gb = cross_val_score(pipeline_gb, X_clean, y_clean, cv=cv, scoring='accuracy')
print(f"Gradient Boosting -> Rata-rata Akurasi: {scores_gb.mean()*100:.2f}% (Std: {scores_gb.std()*100:.2f}%)")
accuracy['Gradient Boosting (CV)'] = scores_gb.mean() * 100
```
